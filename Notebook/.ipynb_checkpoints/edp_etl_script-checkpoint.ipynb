```pyspark
# ---------------------------
# Notebook parameters (edit only this cell)
# ---------------------------
SOURCE_BUCKET = "aws-edp-bucket-source"
DATALAKE_BUCKET = "aws-edp-bucket-datalake"
LOGS_BUCKET = "aws-edp-bucket-emr-logs"
GLUE_DB = "edp_db"

# Which sources to run: choose any subset from ["customer","shopify","health"]
RUN_SOURCES = ["shopify"]   # or ["customer"] etc.

# Which layers to run: choose any subset from ["bronze","silver","gold"]
# You can set to ["bronze","silver","gold"] or ["bronze"] etc.
RUN_LAYERS = ["bronze","silver","gold"]

# Optional: fixed ingestion date for bronze (YYYY-MM-DD) or None to use UTC today
INGEST_DT = None  # e.g. "2025-11-28"

# Spark tuning (tweak if you change instance types)
SPARK_CONF = {
    "spark.sql.shuffle.partitions": "200",
    "spark.executor.instances": "2",
    "spark.executor.cores": "2",
    "spark.executor.memory": "4g",
    "spark.driver.memory": "2g",
    # S3 committer (EMR friendly)
    "spark.hadoop.mapreduce.fileoutputcommitter.algorithm.version": "2",
    "spark.hadoop.fs.s3a.committer.name": "directory",
    "spark.sql.files.maxPartitionBytes": str(128 * 1024 * 1024),
}

print("Parameters set:")
print("Sources:", RUN_SOURCES)
print("Layers:", RUN_LAYERS)
print("Source bucket:", SOURCE_BUCKET)
print("Datalake bucket:", DATALAKE_BUCKET)
print("Ingest date:", INGEST_DT)
```

```pyspark
# ---------------------------
# Imports & Spark builder
# ---------------------------
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_date, to_timestamp, col, year, month, sum as _sum, count
from datetime import datetime
import time

def create_spark(app_name="notebook_edp_etl", configs=None):
    builder = SparkSession.builder.appName(app_name)
    # Glue Catalog integration (optional)
    builder = builder.config(
        "hive.metastore.client.factory.class",
        "com.amazonaws.glue.catalog.metastore.AWSGlueDataCatalogHiveClientFactory"
    )
    defaults = {
        "spark.sql.shuffle.partitions": "200",
        "spark.executor.instances": "2",
        "spark.executor.cores": "2",
        "spark.executor.memory": "4g",
        "spark.driver.memory": "2g",
        "spark.hadoop.mapreduce.fileoutputcommitter.algorithm.version": "2",
        "spark.hadoop.fs.s3a.committer.name": "directory",
        "spark.sql.files.maxPartitionBytes": str(128 * 1024 * 1024),
    }
    if configs:
        defaults.update(configs)
    for k, v in defaults.items():
        builder = builder.config(k, v)
    spark = builder.enableHiveSupport().getOrCreate()
    return spark

# Create spark session (uses SPARK_CONF from params cell)
spark = create_spark("emr_studio_etl", configs=SPARK_CONF)
print("Spark session started:", spark)
print("Spark UI:", spark.sparkContext.uiWebUrl if spark.sparkContext else "N/A")
```

```pyspark
# ---------------------------
# Path helper functions
# ---------------------------
def bronze_path(source_name):
    return f"s3://{DATALAKE_BUCKET}/bronze/{source_name}/"

def silver_path(source_name):
    return f"s3://{DATALAKE_BUCKET}/silver/{source_name}/"

def gold_path(source_name):
    return f"s3://{DATALAKE_BUCKET}/gold/{source_name}/"

def source_path(source_name):
    mapping = {
        "customer": f"s3://{SOURCE_BUCKET}/CustomerData/",
        "shopify":  f"s3://{SOURCE_BUCKET}/Shopify_Data/",
        "health":   f"s3://{SOURCE_BUCKET}/health-violation/",
    }
    return mapping.get(source_name)
```

```pyspark
# ---------------------------
# Bronze ingestion: read raw and write parquet to bronze partition by date
# ---------------------------
def run_bronze(source_name, ingest_dt=None):
    ingest_dt = ingest_dt or datetime.utcnow().strftime("%Y-%m-%d")
    src = source_path(source_name)
    if src is None:
        raise ValueError(f"Unknown source: {source_name}")
    print(f"[BRONZE] Source: {src}")

    # default CSV with header
    df = spark.read.option("header", "true").csv(src)

    dest = bronze_path(source_name) + f"dt={ingest_dt}/"
    print(f"[BRONZE] Writing to: {dest}  (mode=append)")
    start = time.time()
    df.write.mode("append").parquet(dest)
    duration = time.time() - start
    count = df.count()
    print(f"[BRONZE] Completed: rows={count} time={duration:.1f}s")
    return dest
```

```pyspark
# ---------------------------
# Silver transform: source-specific cleaning, typing, dedupe, partitioning
# ---------------------------
def run_silver(source_name):
    bronze_loc = bronze_path(source_name)
    silver_loc = silver_path(source_name)
    print(f"[SILVER] Reading bronze from: {bronze_loc}")
    # WARNING: reading full bronze may be expensive; for prod filter partitions/manifests
    df = spark.read.parquet(bronze_loc)

    if source_name == "customer":
        df2 = (df.withColumn("order_date", to_date(col("order_ts")))
                 .withColumn("amount", col("amount").cast("double"))
                 .dropDuplicates(["order_id"])
                 .na.drop(subset=["order_id"]))
        part_col = "order_date"

    elif source_name == "shopify":
        df2 = (df.withColumn("created_at_ts", to_timestamp(col("created_at")))
                 .withColumn("orders_count", col("orders_count").cast("double"))
                 .dropDuplicates(["customer_id"])
                 .na.fill({"orders_count": 0}))
        df2 = df2.withColumn("created_date", to_date(col("created_at_ts")))
        part_col = "created_date"

    elif source_name == "health":
        df2 = (df.withColumn("inspection_date", to_date(col("inspection_ts")))
                 .withColumn("violation_type", col("violation_type"))
                 .dropDuplicates(["establishment_id", "inspection_id"])
                 .na.drop(subset=["establishment_id"]))
        part_col = "inspection_date"

    else:
        raise ValueError(f"Unknown source for silver: {source_name}")

    print(f"[SILVER] Writing to: {silver_loc} (mode=overwrite partitioned by {part_col})")
    start = time.time()
    # Overwrite for demo idempotency; production: use merge/partition-aware writes
    df2.write.mode("overwrite").partitionBy(part_col).parquet(silver_loc)
    duration = time.time() - start
    cnt = df2.count()
    print(f"[SILVER] Completed: rows={cnt} time={duration:.1f}s")
    return silver_loc
```

```pyspark
# ---------------------------
# Gold aggregation: business-ready tables
# ---------------------------
def run_gold(source_name):
    silver_loc = silver_path(source_name)
    gold_loc = gold_path(source_name)
    print(f"[GOLD] Reading silver from: {silver_loc}")
    df = spark.read.parquet(silver_loc)

    if source_name == "customer":
        agg = df.groupBy(year("order_date").alias("yr"), month("order_date").alias("mon")).agg(_sum("amount").alias("total_amount"))
        part_cols = ["yr", "mon"]

    elif source_name == "shopify":
        if 'created_date' not in df.columns:
            df = df.withColumn('created_date', to_date(col('created_at_ts')))
        agg = df.groupBy(year("created_date").alias("yr"), month("created_date").alias("mon")).agg(_sum("orders_count").alias("total_revenue"))
        part_cols = ["yr", "mon"]

    elif source_name == "health":
        agg = df.groupBy(year("inspection_date").alias("yr"), month("inspection_date").alias("mon")).agg(count("violation_type").alias("total_violations"))
        part_cols = ["yr", "mon"]

    else:
        raise ValueError(f"Unknown source for gold: {source_name}")

    print(f"[GOLD] Writing to: {gold_loc} (mode=overwrite, partitioned by {part_cols})")
    start = time.time()
    agg.write.mode("overwrite").partitionBy(*part_cols).parquet(gold_loc)
    duration = time.time() - start
    cnt = agg.count()
    print(f"[GOLD] Completed: rows={cnt} time={duration:.1f}s")
    return gold_loc
```

```pyspark
# ---------------------------
# Orchestrate runs using RUN_SOURCES and RUN_LAYERS from the params cell
# ---------------------------
print("Starting orchestration")
print("Sources:", RUN_SOURCES)
print("Layers:", RUN_LAYERS)

# Validate
valid_sources = {"customer","shopify","health"}
valid_layers = {"bronze","silver","gold"}

for s in RUN_SOURCES:
    if s not in valid_sources:
        raise ValueError(f"Invalid source: {s}")

for l in RUN_LAYERS:
    if l not in valid_layers:
        raise ValueError(f"Invalid layer: {l}")

# Run in order: bronze -> silver -> gold if present in RUN_LAYERS
layer_order = ["bronze","silver","gold"]
for s in RUN_SOURCES:
    print(f"\n=== Processing source: {s} ===")
    for l in layer_order:
        if l in RUN_LAYERS:
            print(f"> Running layer: {l}")
            if l == "bronze":
                run_bronze(s, ingest_dt=INGEST_DT)
            elif l == "silver":
                run_silver(s)
            elif l == "gold":
                run_gold(s)
print("\nAll requested jobs completed.")
```
